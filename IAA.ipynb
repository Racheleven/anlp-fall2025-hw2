```python
import pandas as pd
import krippendorff

# 读取 CSV
df = pd.read_csv("kappa_high_agreement.csv")  # 替换为你的文件名

# 取出标注者列
annotations = df[['A', 'B', 'C']]

# 转置，Krippendorff 要求 shape = (标注者, 数据条目)
data = annotations.T.values

# 计算 Krippendorff's alpha
alpha = krippendorff.alpha(reliability_data=data, level_of_measurement='nominal')

print(f"Krippendorff's alpha: {alpha:.4f}")
```
