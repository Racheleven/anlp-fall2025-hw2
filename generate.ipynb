```python
import pandas as pd

# 读取 CSV 文件，注意你的 CSV 是 tab 分隔
df = pd.read_csv('1056.csv')

# 只保留 question 列，并写入 questions.txt，每行一个问题
df['question'].to_csv('questions.txt', index=False, header=False)
```

```python
# 构造 reference_answers 字典
import json

reference_answers = {str(i): [ans] for i, ans in enumerate(df['reference_answer'])}

# 写入 JSON 文件
with open('reference_answers.json', 'w', encoding='utf-8') as f:
    json.dump(reference_answers, f, ensure_ascii=False, indent=2)
```
