## Data Cleaning
You need to clean the rows and columns of data using Pandas.  Insert the code (and descriptions of the code in markdown) in this subsection.

```python
df_cars_no_missing = df_cars.dropna()
df_cars_no_missing.info()
```

```python
df_cars.duplicated().any()
```

```python
df_cars.head()
```

```python
if 'Car Make' in df_cars.columns and 'Car Model' in df_cars.columns and 'VIN#' in df_cars.columns:
    df_cars = df_cars.drop(columns=['Car Make','Car Model','VIN#'])
```

```python
df_cars.head()
```

```python
if 'Price' in df_cars:
    df_cars['Price'] = df_cars['Price'].replace(r'[^0-9]','',regex=True)
    df_cars['Price'] = pd.to_numeric(df_cars['Price'],errors='ignore')
    df_cars['Price'] = df_cars['Price'].astype(float)
```

```python
if 'Mileage' in df_cars:
    df_cars['Mileage'] = df_cars['Mileage'].replace(r'[^0-9]','',regex=True)
    df_cars['Mileage'] = pd.to_numeric(df_cars['Mileage'],errors='coerce')
    df_cars = df_cars.dropna(subset=['Mileage'])
    df_cars['Mileage'] = df_cars['Mileage'].astype(float)
```

```python
if 'Doors' in df_cars:
    df_cars['Doors'] = df_cars['Doors'].replace(r'[^0-9]','',regex=True)
    df_cars['Doors'] = pd.to_numeric(df_cars['Doors'],errors='ignore')
    df_cars = df_cars.dropna(subset=['Doors']) 
    df_cars['Doors'] = df_cars['Doors'].astype(int)
```

```python
if 'Year' in df_cars:
    df_cars['Year'] = df_cars['Year'].replace(r'[^0-9]','',regex=True)
    df_cars['Year'] = pd.to_numeric(df_cars['Year'],errors='ignore')
    df_cars = df_cars.dropna(subset=['Year']) 
    df_cars['Year'] = df_cars['Year'].astype(int)
    df_cars = df_cars[(df_cars['Year'] >= 2008) & (df_cars['Year'] <= 2025)]
```

```python
if 'Dealer or Individual?' in df_cars:
    col = 'Dealer or Individual?'
    df_cars[col] = df_cars[col].astype(str).str.strip().str.lower()
    invalid_rows = df_cars[~df_cars[col].isin(['dealer', 'individual'])]
    print("invalid rows:", len(invalid_rows))
    display(invalid_rows)
    mapping = {'dealer': 0, 'individual': 1}
    df_cars[col] = df_cars[col].map(mapping).fillna(2).astype('Int64')
```

```python
df_cars = df_cars.dropna(subset=['Mileage']) 
```

```python
alt.Chart(df_cars, title="Distribution of Car VS Years").mark_bar().encode(
    alt.X("Year:O").title("Years"),
    alt.Y("count()").title("Number of Cars")
).configure_axisX(
    labelAngle=45
)
```

```python
alt.Chart(df_cars, title="Distribution of Cars VS Mileage").mark_bar().encode(
    alt.X("Mileage:O", bin=alt.Bin(maxbins=30)).title("Mileage Range"),
    alt.Y("count()").title("Number of Cars")
).configure_axisX(
    labelAngle=45
)
```

```python
alt.Chart(df_cars, title= "Price VS Year").mark_circle().encode(
    alt.X("Year", scale=alt.Scale(domain=[2008, 2025])).title("Year"),
    alt.Y("Price").title("Price")
)
```

```python
alt.Chart(df_cars, title="Price vs Mileage").mark_circle().encode(
    alt.X("Mileage").title("Mileage"),
    alt.Y("Price").title("Price")
)
```

Some insights:

1. Most car owners sell their manuals through dealers;
2. Most of the used cars for sale are four-door;
3. The price of second-hand cars generally shows a positive correlation with the year of sale. Theoretically, the closer the year is to the present, the less the car has been used, and the higher the selling price;
4. The price of used cars is generally negatively correlated with mileage. The higher the mileage, the lower the selling price of the car.
